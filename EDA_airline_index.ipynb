HISTOGRAM OF 'Fligt Distance', 'Age', 'Departure Delay in Minutes', 'Arrival Delay in Minutes'

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Univariate Analysis: Histograms for 'Fligt Distance', 'Age', 'Departure Delay in Minutes', 'Arrival Delay in Minutes'
fig, axes = plt.subplots(2, 2, figsize=(18, 10))

#axes: coordinates of a plot
#kde = Kernel Density Estimate: estimate the probability density function
#bins = 30: divide data into 30 equal-width intervals

sns.histplot(df['Flight Distance'], bins=30, kde=True, ax=axes[0,0])
axes[0,0].set_title('Histogram of Flight Distance')

sns.histplot(df['Age'], bins=30, kde=True, ax=axes[0,1])
axes[0,1].set_title('Histogram of Age')

sns.histplot(df['Departure Delay in Minutes'], bins=60, kde=True, ax=axes[1,0])
axes[1,0].set_title('Histogram of Departure Delay')

sns.histplot(df['Arrival Delay in Minutes'], bins=60, kde=True, ax=axes[1,1])
axes[1,1].set_title('Histogram of Arrival Delay in Minutes')

plt.show()
```

Analyzing the Impact of Service Features on Satisfaction

```python
service_cols = [
    "Seat comfort",
    "Food and drink",
    "Inflight wifi service",
    "Inflight entertainment",
    "Online support",
    "Ease of Online booking",
    "On-board service",
    "Leg room service",
    "Baggage handling",
    "Checkin service",
    "Cleanliness",
    "Online boarding"
]

import math

n = len(service_cols)
ncols = 4
nrows = math.ceil(n / ncols)

fig, axs = plt.subplots(nrows, ncols, figsize=(16, 3 * nrows))

for i, col in enumerate(service_cols):
    row = i // ncols
    col_idx = i % ncols
    ax = axs[row, col_idx] if nrows > 1 else axs[col_idx]
    sns.boxplot(data=df, x='satisfaction', y=col, ax=ax)
    ax.set_title(f'{col} by Satisfaction')
    ax.set_xlabel('Satisfaction')
    ax.set_ylabel(col)

plt.tight_layout()
plt.show()
```

Impact of Flight Delay on Satisfaction

```python
# The average delay for each satisfaction group
delay_cols = ['Departure Delay in Minutes', 'Arrival Delay in Minutes']
mean_delay = df.groupby('satisfaction')[delay_cols].mean()
print(mean_delay)

plt.figure(figsize=(10,4))
for i, col in enumerate(delay_cols):
    plt.subplot(1, 2, i+1)
    sns.boxplot(data=df, x='satisfaction', y=col)
    plt.title(f'{col} by Satisfaction')
    plt.xlabel('Satisfaction')
    plt.ylabel('Delay (Minutes)')
plt.tight_layout()
plt.show()
print()

# Encode satisfaction: satisfied=1, dissatisfied=0
df['satisfaction_encoded'] = df['satisfaction'].map({'satisfied': 1, 'dissatisfied': 0})

plt.figure(figsize=(8,4))
plt.scatter(df['Departure Delay in Minutes'], df['satisfaction_encoded'], alpha=0.2, label='Departure')
plt.scatter(df['Arrival Delay in Minutes'], df['satisfaction_encoded'], alpha=0.2, label='Arrival')
plt.xlabel('Delay (Minutes)')
plt.ylabel('Satisfaction (0=dissatisfied, 1=satisfied)')
plt.title('Satisfaction vs. Delay')
plt.legend()
plt.tight_layout()
plt.show()

# Bin the departure delays
df['delay_group'] = pd.cut(df['Departure Delay in Minutes'], bins=[-1, 0, 15, 60, 180, 10000],
                           labels=['On time', 'Delay <15m', 'Delay <1h', 'Delay <3h', 'Delay >3h'])
delay_satisfaction = pd.crosstab(df['delay_group'], df['satisfaction'], normalize='index') * 100
print(delay_satisfaction)

delay_satisfaction.plot(kind='bar', stacked=True, figsize=(8,5), colormap='Set2')
plt.title('Satisfaction Rate by Departure Delay Group')
plt.xlabel('Departure Delay Group')
plt.ylabel('Percentage (%)')
plt.legend(title='Satisfaction')
plt.tight_layout()
plt.show()
```
